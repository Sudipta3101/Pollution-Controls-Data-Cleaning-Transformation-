# Detailed Data Cleaning Scenarios (15–20)

## 1. Datetime normalization (24:10 → next day 00:10).

```python
df['TS'].head()
```

```python
df['TS'] = df['TS'].astype(str).str.strip()

mask_24 = df['TS'].str.contains(' 24:', na=False)
df.loc[mask_24, 'TS'] = df.loc[mask_24, 'TS'].str.replace(' 24:', ' 00:')

df['TS'] = pd.to_datetime(df['TS'], format='%d-%m-%Y %H:%M', errors='coerce')

df.loc[mask_24, 'TS'] = df.loc[mask_24, 'TS'] + pd.Timedelta(days=1)
```

```python
df['TS'].head()
```

## 2. Unit standardization (ug/m3, µg/m³ → ug/m3).

```python
print(df["Unit"].unique())
```

```python
df["Unit"] = (df["Unit"].astype(str).str.strip().replace({
        "Âµg/mÂ³": "ug/m3",
        "UG/M3": "ug/m3",
        "mg/nm3": "mg/Nm3"
    })
)
```

```python
print(df["Unit"].unique())
```

```python
print(df["Unit"].value_counts())
```

## 3. Whitespace trimming and casing standardization for status.

```python
print(df["Status"].unique())
```

```python
df["Status"] = df["Status"].astype(str).str.strip().str.upper()
```

```python
print(df["Status"].unique())
```

```python
df["Status"].value_counts()
```

## 4. Coordinate parsing and range validation.

```python
print(df["LatLon"].head())
```

```python
df[["Latitude", "Longitude"]] = df["LatLon"].str.split(",", expand=True)
df["Latitude"] = df["Latitude"].str.strip()
df["Longitude"] = df["Longitude"].str.strip()

df["Latitude"] = pd.to_numeric(df["Latitude"], errors="coerce")
df["Longitude"] = pd.to_numeric(df["Longitude"], errors="coerce")

df.loc[(df["Latitude"] < -90) | (df["Latitude"] > 90), "Latitude"] = pd.NA
df.loc[(df["Longitude"] < -180) | (df["Longitude"] > 180), "Longitude"] = pd.NA
```

## 8. Outlier filtering using rolling median and Hampel filters.

```python
df["PM2_5"].head(50)
```

```python
rolling_median = df["PM2_5"].rolling(window=5, center=True).median()
mad = (df["PM2_5"] - rolling_median).abs().rolling(window=5, center=True).median()
threshold = 3 * 1.4826 * mad
outliers = (df["PM2_5"] - rolling_median).abs() > threshold
df.loc[outliers, "PM2_5"] = rolling_median[outliers]
```

```python
df["PM2_5"].head(50)
```

```python
print("\nOutliers Corrected:", outliers.sum())
```
